## **OPERATORS**

```python
#ASSIGNMENT OPERATORS
a=10
a=a+5
a+=5              #a=a+5
a-=5              #a=a-5
a*=2              #a=a*2
a/=5              #a=a/5
print(a)
```

```python
#CONDITIONAL OPERATORS
m=5
p=10
print(m<p)
print(m>p)
print(m<=p)
print(m>=p)
print(m==p)
print(m!=p)
```

```python
#LOGICAL OPERATORS

age=19
licence='pass'
print((age>18) and (licence=='pass'))
```

```python
#BITWISE OPERATION
x=10    #1010
y=4     #0100

print(x&y)   #and
print(x|y)   #or
print(x^y)   #xor
```

```python
#AREA OF TRIANGLE

base=float(input('Enter a base : '))
height=float(input('Enter a height :'))
area=(0.5*base*height)
print('AREA = ',area)
```

```python
#AREA OF TRIANGLE METHOD 2
import math
a=float(input('length of 1st side : '))
b=float(input('length of 2nd side : '))
c=float(input('length of 3rd side : '))
s=(a+b+c)/2
area=math.sqrt(s*(s-a)*(s-b)*(s-c))
print('AREA =',area)
```

```python
#SOLVING QUADRATIC EQUATION
import math
a=float(input('Enter a : '))
b=float(input('Enter b : '))
c=float(input('Enter c : '))
s=math.sqrt((b*b)-(4*a*c))
x1=(-b+s)/(2*a)
x2=(-b-s)/(2*a)
print("x1 = ",x1 ,"x2 = ",x2)
```

```python
#CONVERT SECOND INTO HR,MIN,SEC
total=int(input('Enter second : '))
hr=total//3600
bal=total%3600
min=bal//60
sec=bal%60
print(hr,'hr',min,'min',sec,'sec')
```

```python
#CONVERT DAYS INTO
total=int(input('Enter total days : '))
year=total//365
bal=(total%365)
month=bal//30
days=bal%30
print(year,'year',month,'month',days,'days')
```

```python
#CONVERT BYTES INTO TB,GB,MB,KB
total=int(input('Enter total bytes : '))
tb=total//(1024**4)
bal1=total%(1024**4)
gb=bal1//(1024**3)
bal2=bal1%(1024**3)
mb=bal2//(1024**2)
bal3=bal2%(1024**2)
kb=bal3//1024
print(tb,'tb',gb,'gb',mb,'mb',kb,'kb')
```
